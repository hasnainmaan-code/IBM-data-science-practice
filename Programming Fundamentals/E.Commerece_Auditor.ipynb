# Module 03: Automated E-Commerce Order Auditor

> **Objective:** Process raw daily order logs, apply conditional volume discounts, calculate sales tax, and output a structured revenue summary payload using parallel loop iteration (`zip()`).

---

### Key Concepts Covered
* Sequence unpacking with `zip()`
* String sanitization & normalization
* Conditional pricing logic (`if/else`)
* Dictionary payload construction & list comprehension aggregations

```python
# Raw order batch stream
order_items = ["  laptop  ", "MOUSE", "  Keyboard  ", "MONITOR"]
unit_prices = [1200.0, 25.0, 75.0, 300.0]
quantities = [1, 4, 2, 1]
```

```python
audited_items=[]
# Step 1: Process transactions via parallel list unpacking
for item,price,qty in zip(order_items,unit_prices,quantities):
    # Normalize product strings
    cleaned_items = item.lower().replace(" ","")
    # Step 2: Apply 10% volume discount for bulk orders (qty >= 3)
    if(qty >= 3):
        discounted_price = price*0.90
    else:
        discounted_price = price
    # Step 3: Calculate subtotal and apply 5% sales tax
    total_price = discounted_price * qty
    final_price = total_price * 1.05
    # Step 4: Append structured record
    audited_items.append({"Item":cleaned_items,"Quantities":qty,"Final Cost":final_price})
    # Step 5: Compute total revenue summary
total_revenue = sum(record["Final Cost"] for record in audited_items)
order ={
     "Audited_items":audited_items,
    "Total_Revenue":total_revenue
    }
print(order)
```
