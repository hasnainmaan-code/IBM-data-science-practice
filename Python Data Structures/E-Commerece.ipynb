# E-Commerce Transaction Stream Ingestion

## Project Overview
This notebook demonstrates a data ingestion pipeline that processes raw, unformatted e-commerce transaction logs into clean, structured Python dictionaries ready for database storage.

* **Concepts Applied:** Tuple Unpacking, String Sanitization, Dictionary Manipulation, Conditional Aggregations.
* **Environment:** Python 3.13.4

### 1. Batch Metadata Extraction

```python
# Step 1: Unpack immutable environment metadata.  
system_metadata = ("APAC_REGION", "2026-Q3", "EUR")
region,quarter,currency = system_metadata
region = region
quarter = quarter
currency = currency
batch_info = {"region":region,"quarter":quarter,"currency":currency} # Dictionary containing batch metadata
```

### 2. Raw Stream Cleaning & Parsing

```python
# Step 2: Sanitize raw log stream and split into individual records
raw_stream_data = "  TXID:9942A#item:Premium_Laptop#price:1249.99#qty:1#status:SUCCESS ;; TXID:9942B#item:Wireless_Mouse#price:45.50#qty:2#status:FAILED  "
stream_data = raw_stream_data.replace(" ","").replace(";;","|")
stream_list = stream_data.split("|") 
```

### 3. Record Construction

```python
# Step 3a: Parse Laptop key-value string into structured dictionary
laptop_str = stream_list[0]
laptop_dict = {} 
for item1 in laptop_str.split("#"):
    key , value = item1.split(":")
    laptop_dict[key] = value
laptop_dict ["price"] = 1249.99
laptop_dict["qty"] = 1
laptop_dict["item"] = "Premium laptop"

# Step 3b: Parse Laptop key-value string into structured dictionary
mouse_str = stream_list[1]
mouse_dict = {}
for item2 in mouse_str.split("#"):
    key , value = item2.split(":")
    mouse_dict[key] = value
mouse_dict ["price"] = 45.50
mouse_dict["qty"] = 2
mouse_dict["item"] = "Wireless mouse"
processed_records = [laptop_dict,mouse_dict]
```

### 4. Metrics & Payload Assembly

```python
# Step 4: Compute summary metrics (maximum unit price & total quantity)
if(laptop_dict["price"]>mouse_dict["price"]):
    max_unit_price = laptop_dict["price"]
else:
    max_unit_price = mouse_dict["price"]
total_quantitiy_processed = mouse_dict["qty"] + laptop_dict["qty"]

metrics = {"total quantitt processed":total_quantitiy_processed,"max unit price":max_unit_price}

final_payload = {"batch_info":batch_info,"processed_records":processed_records,"metrics":metrics}
print(final_payload)


```
