# Task 1: Client Email Parser

> **Objective:** Clean padded whitespace, standardize email strings to uniform lowercase, and split raw data into a clean list using custom delimiters.

```python
raw_emails = "  user1@gmail.com ; USER2@yahoo.com;  user3@outlook.com  "
emails1 = raw_emails.strip().lower().replace(" ","")
emails2 = emails1.split(";")
print(emails2)
```

# Task 2:Financial Portfolio Slice

> **Objective:** Apply positive and negative list slicing to extract daily stock sub-periods, and calculate statistical aggregations using built-in functions (`max()`, `min()`, `sum()`).

```python
stock_prices = [150.2, 152.4, 148.9, 155.1, 158.0, 156.3, 160.1, 159.4, 162.8, 165.0]
print(f"The prices in first five days are {stock_prices[:5]}")
print(f"The stock prices in last three days are {stock_prices[-3:]}")
print(f"The maximum price is {max(stock_prices)}\nThe minimum prices are {min(stock_prices)}\nThe total sum of all prices are {sum(stock_prices)} ")
```

# Task 5:Structured Data Parsing

> **Objective:** Parse raw piped sensor string logs into a structured Python key-value dictionary using loop iteration and string splitting.

Structured Data Parsing

```python
sensor_log = "device_id:SENS-901|temp:24.5|humidity:60%"
readings = sensor_log.split("|")
Sensor_Data = {}
for item in readings:
    key, value = item.split(":")
    Sensor_Data[key] = value
print(Sensor_Data)

```
